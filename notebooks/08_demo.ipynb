### Варіант B — інтерактивна форма (якщо встановлено `ipywidgets`)

```python
try:
    import ipywidgets as W
    from IPython.display import display
    cap=W.Textarea(description='Підпис:'); dur=W.IntText(value=20, description='Трив., с:')
    when_=W.Text(value='2025-03-01T19:30:00', description='Час UTC:'); cr=W.Text(description='Креатор:')
    out=W.Output(); btn=W.Button(description='Передбачити', button_style='primary')
    def go(_):
        out.clear_output()
        with out:
            r=inference.recommend(caption=cap.value, duration=dur.value or None, when=when_.value or None, creator=cr.value or None)
            print(r['recommendation'], r['probability']); print(r['decision_reason'])
            print('фактори:', [f['label'] for f in r['factors'][:3]])
    btn.on_click(go); display(W.VBox([cap,dur,when_,cr,btn,out]))
except ImportError:
    print('ipywidgets не встановлено — користуйся Варіантом A (pip install ipywidgets)')
```
